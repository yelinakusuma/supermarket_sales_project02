```python
import warnings
warnings.filterwarnings('ignore')
from great_expectations.data_context import FileDataContext

context = FileDataContext.create(project_root_dir = './')
```

## **Connect to a Datasource**

Bagian ini dilakukan untuk mendefinisikan sumber data yang akan digunakan dalam proses validasi.

```python
# Membuat datasource: untuk mendaftarkan sumber data ke dalam great expectations
datasource_name = 'csv_files'
datasource = context.sources.add_pandas(datasource_name)

# Membuat data asset: untuk mendefinisikan data asset yang berasal dari file CSV
asset_name = 'cleaned_data'
csv_path = '/content/drive/MyDrive/P2M3_yelina_kusuma_data_clean.csv'
asset = datasource.add_csv_asset(asset_name, filepath_or_buffer = csv_path)

# Membuat batch request: untuk mengakses data dalam bentuk batch yang akan digunakan saat membuat validator
batch_request = asset.build_batch_request()
```

## **Create an Expectation Suite**

Bagian ini digunakan untuk mendefinisikan atau membuat wadah berisi aturan-aturan validasi data yang akan digunakan dalam proses selanjutnya.

```python
# Membuat expectation suite
expectation_suite_name = 'sales-report'
context.add_or_update_expectation_suite(expectation_suite_name)

# Membuat validator
validator = context.get_validator(
    batch_request = batch_request,
    expectation_suite_name = expectation_suite_name
)

# Cek validator
validator.head()
```

### 1. to be unique

Fungsi ini digunakan untuk memastikan bahwa setiap nilai di sutau kolom tidak ada yang duplikat. Pada project ini diterapkan pada kolom `invoice_id` yang mana nilainya harus bersifat unique.

```python
validator.expect_column_values_to_be_unique("invoice_id")
```

### 2. to be between min_value and max_value

Fungsi ini digunakan untuk memastikan bahwa nilai-nilai di kolom berada di rentang tertentu. Pada project ini diterapkan pada kolom `quantity` yang mana nilai-nilainya harus berada antara 1 hingga 10.

```python
validator.expect_column_values_to_be_between("quantity", min_value = 1, max_value = 10)
```

### 3. to be in set

Fungsi ini digunakan untuk memastikan bahwa nilai-nilai pada suatu kolom hanya berasal dari sekumpulan nilai yang sudah ditentukan. Pada project ini diterapkan pada kolom `gender` untuk memastikan nilai-nilainya hanya boleh male dan female.

```python
validator.expect_column_values_to_be_in_set("gender", ["Male", "Female"])
```

### 4. to be in type list

Fungsi ini digunakan untuk memastikan bahwa tipe data pada suatu kolom termasuk dalam list tipe data yang sudah ditentukan. Pada project ini diterapkan pada kolom `unit_price` yang mana harus bertipe float atau integer.

```python
validator.expect_column_values_to_be_in_type_list("unit_price", ['float', 'int'])
```

### 5. to match regex

Fungsi ini digunakan untuk memastikan bahwa nilai-nilai pada suatu kolom sesuai dengan pola regex tertentu. Pada project ini diterapkan untuk memastikan bahwa nilai-nilai pada kolom `invoice_id` harus punya format seperti 750-67-8428.

```python
validator.expect_column_values_to_match_regex("invoice_id", r"^\d{3}-\d{2}-\d{4}$")
```

### 6. to be between median

Fungsi ini digunakan untuk memastikan bahwa nilai median pada suatu kolom berada di antara batas minimum dan maksimum yang sudah ditentukan. Pada project ini diterapkan pada kolom `rating` untuk memastikan distribusinya tidak terlalu menyimpang atau bias.

```python
validator.expect_column_median_to_be_between("rating", min_value = 1.0, max_value = 10.0)
```

### 7. to match datetime format

Fungsi ini digunakan untuk memastikan bahwa format tanggal di suatu kolom sesuai dengan format waktu tertentu. Pada project ini diterapkan pada kolom `date` yang mana harus sesuai dengan format "%Y-%m-%d", seperti 2019-01-05.

```python
validator.expect_column_values_to_match_strftime_format("date", "%Y-%m-%d")
```

Berdasarkan ketujuh expectations yang telah diterapkan, seluruhnya berhasil dijalankan dengan status **success: true**. Hal ini menunjukkan bahwa data telah berhasil memenuhi kriteria validasi yang telah ditentukan dan dapat digunakan untuk tahap analisis data atau pemodelan selanjutnya.
